```r
x = 2:12
p = c( 0, 2, 3, 4, 5, 8, 5, 4, 3, 2, 0 ) / 36
names(p) = sprintf( "x=%d", x )
sum(p)
print(x)
print(p)
barplot(p)
```

```r
mean = sum( p * x )

variance = sum( p * ( x - mean ) ^2 )

print( sprintf( "The mean is %.2f!", mean ))
print( sprintf( "The variance is %.2f!", variance ))

print( sprintf( "The standard deviation is %.2f!", sqrt(variance) ))
```

```r
my_data = c( 1, 2, 3, 5, 4, 6, 3, 7, 1, 2, 5, 4, 6, 7 )
mean(my_data)
var(my_data)
sd(my_data)
```
